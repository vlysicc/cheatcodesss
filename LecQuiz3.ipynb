STEP 2

```python

display(df[df['patient_name'].isna()])
```

STEP 3

```python

print("Before cleaning:", df['barangay'].nunique())

df['barangay'] = df['barangay'].str.strip().str.title()

print("After cleaning:", df['barangay'].nunique())
print(df['barangay'].unique())
```

STEP 4

```python

df['visit_date'] = pd.to_datetime(df['visit_date'], errors='coerce')

df = df.sort_values('visit_date')

display(df[['patient_name', 'visit_date']])

print("\nVisits per day:")
print(df['visit_date'].value_counts().sort_index())
```

STEP 5

```python

display(
    df[df['patient_name'].duplicated(keep=False)]
    [['patient_name', 'age', 'barangay', 'visit_date']]
)
```

STEP 6

```python

df['patient_name'] = df['patient_name'].str.strip().str.title()

df['sex'] = df['sex'].str.strip().str.lower()
df['sex'] = df['sex'].replace({
    'm': 'Male',
    'male': 'Male',
    'f': 'Female',
    'female': 'Female'
})

df['diagnosis'] = df['diagnosis'].str.strip().str.title()
df['diagnosis'] = df['diagnosis'].replace({
    'Migrain': 'Migraine'
})

df['follow_up_needed'] = df['follow_up_needed'].str.strip().str.lower()
df['follow_up_needed'] = df['follow_up_needed'].replace({
    'yes': 'Yes',
    'y': 'Yes',
    'no': 'No',
    'n': 'No'
})


df.loc[(df['age'] < 0) | (df['age'] > 120), 'age'] = np.nan

df_clean = df.dropna(subset=['patient_name']).copy()

display(df_clean)

print("Total valid patients:", len(df_clean))
print("Average age:", round(df_clean['age'].mean(), 2))
print("Most common diagnosis:", df_clean['diagnosis'].mode()[0])

follow_up = (df_clean['follow_up_needed'] == 'Yes').mean() * 100
print("Percentage needing follow-up:", round(follow_up, 2), "%")
```
