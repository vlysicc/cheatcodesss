**STEP 1**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

data = {
    'student_name' : ['Miguel', 'Andrea', 'Josh', 'Kyla', 'Ranier', 'Bea', 'Carlo', 'Denise', 'Enzo', 'Fiona', 'Gab', 'Hazel', 'Ivan', 'Jelly', 'Kurt', 'Liza', 'Mark', 'Nina', 'Oscar', 'Pia' ],
    'study_hours' : [1,2,3,4,5,6,6,7,8,8,9,10,1,2,3,3,5,6,7,7],
    'attendance_rate' : [60,65,70,75,80,78,85,88,90,92,95,93,97,99,55,62,72,68,82,90],


}

df = pd.DataFrame(data)
print(df[["student_name", "study_hours", "attendance_rate"]])
```

**STEP 2**

```python
from sklearn.preprocessing import StandardScaler

X = df[['study_hours', 'attendance_rate']]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
print(X_scaled[:5])
```

**STEP 3**

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X_scaled)
print(df[["student_name", "study_hours", "attendance_rate", "cluster"]])
```

**STEP 4**

```python
centers_original = scaler.inverse_transform(kmeans.cluster_centers_)
for i, c in enumerate(centers_original):
  print(f"cluster {i}: study hours = {c[0]:.1f}, Attendace rate = {c[1]:.1f}")
```
